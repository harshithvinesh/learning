<p>Problem Statement: An aerospace engineer is designing the fuel-line distribution tree for a commercial jet. The central fuselage tank is the root, branching into the left wing (root.left) and right wing (root.right). Each sub-tank stores a specific fuel capacity (integer value), and -1 represents a sealed/empty valve. For the aircraft to maintain aerodynamic balance during flight, the left wing's tank network must be an exact structural and capacity mirror of the right wing. Return true if the aircraft is safe to fly, or false otherwise.

Key Exam Clue: "Left wing must be a mirror image of the right wing."</p>

```
import java.util.*;

class TreeNode {
    int val;
    TreeNode left_wing, right_wing;
    TreeNode(int val) {
        this.val = val;
    }
}

public class Solution {
    public static TreeNode buildEngine(String[] nodes) {
        if (nodes.length == 0 || nodes[0].equals("-1")) return null;
        TreeNode root = new TreeNode(Integer.parseInt(nodes[0]));
        Queue<TreeNode> q = new LinkedList<>();
        q.add(root);
        
        // FIX 1: Start at index 1 (since index 0 is already used for root!)
        int seat = 1; 
        
        while (!q.isEmpty() && seat < nodes.length) {
            TreeNode curr = q.poll();
            if (!nodes[seat].equals("-1")) {
                curr.left_wing = new TreeNode(Integer.parseInt(nodes[seat]));
                q.add(curr.left_wing);
            }
            seat++;
            if (seat < nodes.length && !nodes[seat].equals("-1")) {
                curr.right_wing = new TreeNode(Integer.parseInt(nodes[seat]));
                q.add(curr.right_wing);
            }
            seat++;
        }
        return root;
    }

    // FIX 2 & 3: Placed inside Solution class & updated to use left_wing / right_wing
    public static boolean isSymmetric(TreeNode root) {
        if (root == null) return true;
        return isMirror(root.left_wing, root.right_wing);
    }

    private static boolean isMirror(TreeNode t1, TreeNode t2) {
        if (t1 == null && t2 == null) return true;
        if (t1 == null || t2 == null) return false;
        return (t1.val == t2.val) 
            && isMirror(t1.left_wing, t2.right_wing)   // Outer wing tanks must match
            && isMirror(t1.right_wing, t2.left_wing);  // Inner wing tanks must match
    }
}

// --- JUPYTER TEST RUNNER ---

// Example 1: Safe to fly (Outer tanks = 80, Inner tanks = 120)
String safeJet = "500 200 200 80 120 120 80";
TreeNode jetRoot1 = Solution.buildEngine(safeJet.split(" "));
System.out.println("Jet 1 Safe to Fly: " + Solution.isSymmetric(jetRoot1)); // Prints: true

// Example 2: Unsafe! Left wing has [80, 120] and Right wing also has [80, 120] (Not mirrored!)
String unsafeJet = "500 200 200 80 120 80 120";
TreeNode jetRoot2 = Solution.buildEngine(unsafeJet.split(" "));
System.out.println("Jet 2 Safe to Fly: " + Solution.isSymmetric(jetRoot2)); // Prints: false

// Example 3: Unsafe! Sealed valve (-1) on the inner right wing
String leakingJet = "500 200 200 80 120 -1 80";
TreeNode jetRoot3 = Solution.buildEngine(leakingJet.split(" "));
System.out.println("Jet 3 Safe to Fly: " + Solution.isSymmetric(jetRoot3)); // Prints: false
```

```
import java.util.*;

class TreeNode {
    int val;
    TreeNode left, right;
    TreeNode(int val) {
        this.val = val;
    }
}

public class Solution {
    // Same buildTree from Question 1
    public static TreeNode buildTree(String[] nodes) {
        if (nodes.length == 0 || nodes[0].equals("-1")) return null;
        TreeNode root = new TreeNode(Integer.parseInt(nodes[0]));
        Queue<TreeNode> q = new LinkedList<>();
        q.add(root);
        int i = 1;
        while (!q.isEmpty() && i < nodes.length) {
            TreeNode curr = q.poll();
            if (!nodes[i].equals("-1")) {
                curr.left = new TreeNode(Integer.parseInt(nodes[i]));
                q.add(curr.left);
            }
            i++;
            if (i < nodes.length && !nodes[i].equals("-1")) {
                curr.right = new TreeNode(Integer.parseInt(nodes[i]));
                q.add(curr.right);
            }
            i++;
        }
        return root;
    }

    // --- QUESTION #2: AVERAGE OF LEVELS IN B-TREE (LC 637) ---
    public static List<Double> averageOfLevels(TreeNode root) {
        List<Double> result = new ArrayList<>();
        if (root == null) return result;

        Queue<TreeNode> q = new LinkedList<>();
        q.add(root); // Start with Level 0 (Root)

        while (!q.isEmpty()) {
            int size = q.size(); // Number of nodes on the CURRENT level
            double sum = 0;      // Use double to avoid overflow & get decimal division

            // Loop through ONLY the nodes on the current level
            for (int i = 0; i < size; i++) {
                TreeNode curr = q.poll();
                sum += curr.val;

                // Push next level's children into the queue for the next while-loop
                if (curr.left != null) q.add(curr.left);
                if (curr.right != null) q.add(curr.right);
            }

            // Current level finished! Calculate average and add to list
            result.add(sum / size);
        }
        return result;
    }
}

// --- JUPYTER TEST RUNNER ---
// Tree Structure:
//         3          (Level 0 -> Avg: 3.0)
//       /   \
//      9     20      (Level 1 -> Avg: (9 + 20) / 2 = 14.5)
//           /  \
//          15   7    (Level 2 -> Avg: (15 + 7) / 2 = 11.0)
TreeNode orgChart = Solution.buildTree("3 9 20 -1 -1 15 7".split(" "));
System.out.println("Average per Tier: " + Solution.averageOfLevels(orgChart)); 
// Prints: [3.0, 14.5, 11.0]
/
d
```

```
import java.util.*;

public class Solution6 {
    // --- QUESTION #6: CLIMBING STAIRS (LC 70) ---
    public static int climbStairs(int n) {
        // Base cases: 1 way for 1 step, 2 ways for 2 steps
        if (n <= 2) return n;

        int prev2 = 1; // Ways to reach (i - 2), starts at Step 1
        int prev1 = 2; // Ways to reach (i - 1), starts at Step 2

        for (int i = 3; i <= n; i++) {
            int curr = prev1 + prev2; // Ways to reach current step i
            prev2 = prev1;            // Shift window forward
            prev1 = curr;
        }
        return prev1;
    }
}

// --- JUPYTER TEST RUNNER ---
System.out.println("Ways for 3 steps: " + Solution6.climbStairs(3)); // Prints: 3
System.out.println("Ways for 5 steps: " + Solution6.climbStairs(5)); // Prints: 8
```

```
import java.util.*;

public class Solution7 {
    // --- QUESTION #7: MIN COST CLIMBING STAIRS (LC 746) ---
    public static int minCostClimbingStairs(int[] cost) {
        int n = cost.length;
        
        int prev2 = cost[0]; // Min cost to step on index 0
        int prev1 = cost[1]; // Min cost to step on index 1

        for (int i = 2; i < n; i++) {
            // Pay current step's cost + cheaper of the previous two steps
            int curr = cost[i] + Math.min(prev1, prev2);
            prev2 = prev1;
            prev1 = curr;
        }

        // Finish line is past the last element! You can jump there from (n-1) or (n-2)
        return Math.min(prev1, prev2);
    }
}

// --- JUPYTER TEST RUNNER ---
int[] towers1 = {10, 15, 20};
System.out.println("Min Toll 1: " + Solution7.minCostClimbingStairs(towers1)); // Prints: 15

int[] towers2 = {1, 100, 1, 1, 1, 100, 1, 1, 100, 1};
System.out.println("Min Toll 2: " + Solution7.minCostClimbingStairs(towers2)); // Prints: 6
```

```
import java.util.*;

public class Solution10 {
    // --- QUESTION #10: BRACE EXPANSION (LC 1087) ---
    public static String[] expand(String s) {
        List<String> result = new ArrayList<>();
        backtrack(s, 0, "", result);
        
        // Exam requirement: always sort the final list alphabetically!
        Collections.sort(result);
        
        // Convert List<String> to String[] (or just return 'result' if portal wants a List)
        return result.toArray(new String[0]);
    }

    private static void backtrack(String s, int idx, String current, List<String> result) {
        // Base case: reached the end of the pattern string
        if (idx == s.length()) {
            result.add(current);
            return;
        }

        // Case 1: We hit a '{' -> branch on all comma-separated options
        if (s.charAt(idx) == '{') {
            int closeIdx = s.indexOf('}', idx); // Find matching '}'
            String[] options = s.substring(idx + 1, closeIdx).split(",");
            
            for (String opt : options) {
                // Jump straight to closeIdx + 1 (past the '}')
                backtrack(s, closeIdx + 1, current + opt, result);
            }
        } 
        // Case 2: Normal letter -> just attach it and move +1
        else {
            backtrack(s, idx + 1, current + s.charAt(idx), result);
        }
    }
}

// --- JUPYTER TEST RUNNER ---
String pattern1 = "{a,b}c{d,e}f";
System.out.println("Expanded 1: " + Arrays.toString(Solution10.expand(pattern1)));
// Prints: [acdf, acef, bcdf, bcef]

String pattern2 = "abcd";
System.out.println("Expanded 2: " + Arrays.toString(Solution10.expand(pattern2)));
// Prints: [abcd]
```

```
import java.util.*;

public class Solution11 {
    // --- QUESTION #11: GRAY CODE (LC 89) ---
    public static List<Integer> grayCode(int n) {
        List<Integer> result = new ArrayList<>();
        int totalNumbers = 1 << n; // Same as (int) Math.pow(2, n)

        for (int i = 0; i < totalNumbers; i++) {
            result.add(i ^ (i >> 1)); // Bitwise XOR with right-shift
        }
        return result;
    }
}

// --- JUPYTER TEST RUNNER ---
System.out.println("Gray Code (n = 2): " + Solution11.grayCode(2)); // Prints: [0, 1, 3, 2]
System.out.println("Gray Code (n = 3): " + Solution11.grayCode(3)); // Prints: [0, 1, 3, 2, 6, 7, 5, 4]
```

```
import java.util.*;

public class Solution12 {
    // --- QUESTION #12: PATH WITH MAXIMUM GOLD (LC 1219) ---
    public static int getMaximumGold(int[][] grid) {
        int maxGold = 0;
        for (int r = 0; r < grid.length; r++) {
            for (int c = 0; c < grid[0].length; c++) {
                if (grid[r][c] > 0) {
                    maxGold = Math.max(maxGold, dfs(grid, r, c));
                }
            }
        }
        return maxGold;
    }

    private static int dfs(int[][] grid, int r, int c) {
        // Boundary check or hitting an obstacle (0)
        if (r < 0 || r >= grid.length || c < 0 || c >= grid[0].length || grid[r][c] == 0) {
            return 0;
        }

        int currentGold = grid[r][c];
        grid[r][c] = 0; // Temporarily mark as visited in-place

        int bestNext = 0;
        // Explore 4 directions: Down, Up, Right, Left
        bestNext = Math.max(bestNext, dfs(grid, r + 1, c));
        bestNext = Math.max(bestNext, dfs(grid, r - 1, c));
        bestNext = Math.max(bestNext, dfs(grid, r, c + 1));
        bestNext = Math.max(bestNext, dfs(grid, r, c - 1));

        grid[r][c] = currentGold; // Backtrack: restore cell value for other paths
        return currentGold + bestNext;
    }
}

// --- JUPYTER TEST RUNNER ---
int[][] goldGrid = {
    {0, 6, 0},
    {5, 8, 7},
    {0, 9, 0}
};
System.out.println("Max Gold Collected: " + Solution12.getMaximumGold(goldGrid)); // Prints: 24 (Path: 9 -> 8 -> 7)
```
